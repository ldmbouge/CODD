```python
benches = {}
with open('all_tsp') as f:
    for line1 in f:
        solver, bench, w, _ = line1.split(' ')
        w = int(w)
        #bench = '/'.join(bench.split('/')[-2:])
        line2 = f.readline()
        if len(line2) > 0 and 'T.O.' not in line2:
            obj = float(line2.split('\t')[0].split(':')[-1])

            if bench not in benches:
                benches[bench] = [(solver, w, obj)]
            else:
                benches[bench].append((solver, w, obj))
```

```python
nGood = 0
for bench, data in benches.items():
    all_obj = data[0][-1] # data[0] is the first, [-1] is the objective 
    all_good = True
    for solver, w, obj in data:
        if obj != all_obj:
            all_good = False
            print(f'./{solver} {bench} {w} -> {obj}, NOT {all_obj}')
    if all_good:
        nGood += 1
        #print(f'all solvers agreed on {"/".join(bench.split("/")[-2:])} at widths={set(x[1] for x in data)}')
print(f'{nGood}/{len(benches.keys())} passed')
```
