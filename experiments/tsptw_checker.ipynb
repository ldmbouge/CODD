```python
def check(P,solution):
    n,a,b,d = P

    if len(solution) != n:
        print(f'solution length is not correct: n = {n}, |solution| = {len(solution)}')

    t = 0
    cost = 0
    curr = 0
    unvisited = list(range(n))
    for next in solution:
        delta_cost = d[curr][next]
        next_t = t + delta_cost
        print(f'{curr: 3} -> {next}\tt={i2d(next_t): 10}\td={i2d(next_t - t): 10}\ttw=[{i2d(a[next]): 10}, {i2d(b[next]): 10}]    cost: {cost+delta_cost}')
        
        for j in range(0, n):
            if j in unvisited and (t + d[curr][j] > b[j]): print(f'AACK! {j}, {curr}, {unvisited}, {t} + {d[curr][j]} = {t + d[curr][j]} > {b[j]}')
        
        if next_t > b[next]:
            print(f'past end of time window')
            return None
        elif next_t < a[next]:
            print(f'waiting for {a[next]}')
            delta = a[next] - next_t
            t += delta
        else:
            #print(f'arrive in window: {i2d(next_t)} in [{i2d(a[prev]): 10}, {i2d(b[prev]): 10}]')
            t = next_t
        cost += delta_cost
        
        unvisited.remove(curr)
        curr = next
    return i2d(cost)
```
