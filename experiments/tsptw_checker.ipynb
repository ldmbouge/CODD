```python
def is_triangular(n,d):
    for i in range(n):
        for j in range(1,n):
            for k in range(n):
                if d[i][j] + d[j][k] < d[i][k]:
                    # print(f'd[{i}][{j}] + d[{j}][{k}] < d[{i}][{k}]')
                    # print(f'{d[i][j]} + {d[j][k]} < {d[i][k]}')
                    # tri=False
                    return False
    return True
```

```python
import re

def d2i(d): 
    return int(float(d) * 10000)
def i2d(i):
    return i / 10000.0

def read_model(fname, dir='../data/tsptw/'):
    with open(f'{dir}{fname}', 'r') as in_f:
        lines = in_f.readlines()

        i = 0
        a,b,c = [],[],[]

        while lines[i].startswith('#'): i += 1
        n = int(lines[i]); i += 1

        while i <= n:
            if lines[i].startswith('#'): 
                i += 1
                continue
            c.append([d2i(x) for x in lines[i].split(' ')]); i += 1

        while i < len(lines):
            if lines[i].startswith('#'): 
                i += 1
                continue
            [_a, _b] = [d2i(x) for x in re.findall(r'(\d+\.?\d*)\s+(\d+\.?\d*)', lines[i])[0]]; i += 1
            a.append(_a)
            b.append(_b)
    #print(n,a,b,c)
    #print(is_triangular(n,c))
    return n,a,b,c
```

```python
def check(P,solution):
    n,a,b,d = P

    if len(solution) != n:
        print(f'solution length is not correct: n = {n}, |solution| = {len(solution)}')

    t = 0
    cost = 0
    curr = 0
    unvisited = list(range(n))
    for next in solution:
        delta_cost = d[curr][next]
        next_t = t + delta_cost
        print(f'{curr: 3} -> {next}\tt={i2d(next_t): 10}\td={i2d(next_t - t): 10}\ttw=[{i2d(a[next]): 10}, {i2d(b[next]): 10}]')
        
        for j in range(0, n):
            if j in unvisited and (t + d[curr][j] > b[j]): print(f'AACK! {j}, {curr}, {unvisited}, {t} + {d[curr][j]} = {t + d[curr][j]} > {b[j]}')
        
        if next_t > b[next]:
            print(f'past end of time window')
            return None
        elif next_t < a[next]:
            print(f'waiting for {a[next]}')
            delta = a[next] - next_t
            t += delta
        else:
            #print(f'arrive in window: {i2d(next_t)} in [{i2d(a[prev]): 10}, {i2d(b[prev]): 10}]')
            t = next_t
        cost += delta_cost
        
        unvisited.remove(curr)
        curr = next
    return i2d(cost)
```
